The dataset contained **506** missing(NaN)values in the **WIP** column.Since WIP is a numerical feature,these missing/NaN entries were replaced using the median,which is more robust to outliers and preserves the overall distribution.This ensures the dataset remains complete and suitable for further preprocessing.


```python
#Displaying the Missing(NaN)Values in the WIP Column
print("Missing(Nan)Values in the WIP Column:",df["wip"].isnull().sum())
```

```python
#Filling WIP missing values with Median
df["wip"]=df["wip"].fillna(df["wip"].median())
```

```python
#After Replacing WIP missing values with Median
print("Missing(NaN)Values in the WIP Column:",df["wip"].isnull().sum())
```

```python
#Dispalying WIP Median Value
print("WIP Median Value:",df["wip"].median())
```

#### (iii) Bivariate Analysis
Bivariate analysis means studying the relationship between two variables at a time, especially how different factors are related to actual_productivity.

##### 1. Numerical vs Numerical

```python
# Correlation of numerical variables with actual productivity

numeric_cols = df.select_dtypes(include='number')

correlation = numeric_cols.corr()['Actual Productivity'].sort_values(
    ascending=False
)

print(correlation)
```

***Info:-*** This tells you how strongly each **numerical variable** is related to productivity.Targeted Productivity has the strongest positive relationship with Actual Productivity **(0.421594)**, indicating a moderate association. Most other numerical variables show weak or very weak relationships with actual productivity. No. of Style Changes (-0.207366) has the strongest negative relationship, suggesting that increased style changes may be associated with slightly lower productivity.

##### 2. Targeted Productivity vs Actual Productivity

***Info:-*** The correlation between Targeted Productivity and **Actual Productivity is 0.421594**The correlation is **positive**, meaning higher targeted productivity is generally associated with higher actual productivity.

#### (iv) Multivariate Analysis
It examines how multiple variables interact simultaneously to uncover deeper patterns,combined effects,and factors influencing the target outcome.

```python
print("Department + Quarter + Actual Productivity")
print(df.groupby(['Department', 'Quarter'])['Actual Productivity'].agg(
    ['count', 'mean', 'min', 'max']
))

print("\nTeam + Department + Actual Productivity")
print(df.groupby(['Team', 'Department'])['Actual Productivity'].agg(
    ['count', 'mean', 'min', 'max']
))

print("\nTeam + Quarter + Actual Productivity")
print(df.groupby(['Team', 'Quarter'])['Actual Productivity'].agg(
    ['count', 'mean', 'min', 'max']
))
```

***Purpose:-*** Multivariate analysis was performed by examining actual productivity across multiple variables simultaneously, including **department, team, and quarter**. Groupby analysis was used to identify how productivity varies across different combinations of these factors. This helped identify differences in productivity performance among teams, departments, and production periods.
