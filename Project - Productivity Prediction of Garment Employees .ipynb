```python
#Set the figure size
plt.figure(figsize=(8, 5))

#Create Box Plot
sns.boxplot(
    data=df,
    x='Department',
    y='Actual Productivity',
    color='hotpink'
)

#Set the plot title
plt.title('Actual Productivity Distribution by Department')

#Label the X-axis and Y-axis
plt.xlabel('Department')
plt.ylabel('Actual Productivity')

#Display the plot
plt.show()
```

### 7 Actual Productivity

• ***Visualization Type:-*** KDE (Kernel Density Estimation) Plot
• ***Purpose:-*** To understand the **shape and density of the actual productivity distribution**, including where productivity values are concentrated.
• ***Insight:-*** The KDE plot shows the productivity ranges where observations are most concentrated. It helps identify the **most common productivity levels**, the overall distribution shape, and whether the data is skewed or has multiple peaks.

```python
#Set the figure size
plt.figure(figsize=(8, 5))

#Create KDE Plot
sns.kdeplot(df['Actual Productivity'], fill=True,alpha=0.6,
    linewidth=1.5,color='green')

#Set the plot title
plt.title('Density Distribution of Actual Productivity')

#Label the X-axis and Y-axis
plt.xlabel('Actual Productivity')
plt.ylabel('Density')

#Display the plot
plt.show()
```

### 8  Team vs Actual Productivity

• ***Visualization Type:-*** Violin Plot
• ***Purpose:-*** To compare the distribution and variability of actual productivity across different teams.
• ***Insight:-*** The plot shows how actual productivity is distributed among the teams. **Team 1 has the highest average productivity (0.821054), while Team 7 has the lowest average productivity (0.668006)**. This indicates that productivity performance varies across teams.

```python
#Set the figure size
plt.figure(figsize=(12, 6))

#Create Violin Plot
sns.violinplot(
    data=df,
    x='Team',
    y='Actual Productivity',
    color='red'
)
#Set the plot title
plt.title('Actual Productivity Distribution by Team')

#Label the X-axis and Y-axis
plt.xlabel('Team')
plt.ylabel('Actual Productivity')

#Display the plot
plt.show()
```

### 9  Correlation Heatmap

• ***Visualization Type:-*** Heatmap
• ***Purpose:-*** Shows the **strength and direction of relationships among numerical variables**.
• ***Insight:-*** **Targeted Productivity has the strongest positive correlation with Actual Productivity (0.421594)**. **No Of Style Change has the strongest negative correlation (-0.207366)**, followed by **Idle Men (-0.181734)**. Most other variables have weak or very weak correlations with actual productivity. This indicates that productivity is influenced by multiple factors rather than one variable alone.

```python
#Set the figure size
plt.figure(figsize=(12, 8))

#select all numerical columns from the DataFrame df.
numeric_cols = df.select_dtypes(include='number')

#Create Heatmap
sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
#Set the plot title
plt.title('Correlation Heatmap of Numerical Variables')

#Display the plot
plt.show()
```

### 10  Productivity by Department and Quarter

• ***Visualization Type:-*** Subplot / Multiple Bar Plots
• ***Purpose:-*** To compare **average actual productivity across different quarters for each department** and identify how productivity varies between departments and production periods.
• ***Insight:-*** Productivity varies across both departments and quarters. The **Finishing departmen**t generally has **higher average productivity than the Sewing departmen**t. Productivity also changes across quarters, indicating that production period and department both influence productivity performance.

```python
#Create Sub Plot
g = sns.catplot(
    data=df,
    x='Quarter',
    y='Actual Productivity',
    col='Department',
    kind='bar',
    height=5,
    aspect=1.2,
    color='red'
)

#Label the X-axis and Y-axis
g.set_axis_labels('Quarter', 'Average Actual Productivity')

#Sets the title of each individual subplot according to the department.
g.set_titles('{col_name} Department')

#Adds one main title above all the subplots.
g.fig.suptitle(
    'Average Actual Productivity by Department and Quarter',
    y=1.05
)

#Display the plot
plt.show()
```
